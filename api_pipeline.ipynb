```python
# transfer data to dataframe
def transform_data (data):
    # extract values
    times_series = data['values']

    # covert to dataframe
    df = pd.DataFrame(times_series)

    # convert to proper data types
    df['datetime'] = pd.to_datetime(df['datetime'])

    df = df.astype({ 
        'open': 'float',
        'high':'float',
        'low': 'float', 
        'close': 'float',
        'volume': 'int'
    })

    return df
```

##  Extract data for multiple symbols

```python
symbols = [ 'AAPL', 'MSFT', 'GOOGL', 'AMZN' # original symbols requested to be extracted
            'TSLA', 'NVDA', 'NFLX', 'META', 'JPM' # 5 add_symbols 
] 

symbols_str = ",".join(symbols)
url = f"https://api.twelvedata.com/time_series?symbol={symbols_str}&interval=1day&apikey={api_key}&outputsize=20"
response = requests.get(url)
data = response.json()

# all_data =[] commented the below out as the addtion request of the new symbols to be added was hittng the API rate limit. The website allows 8 requests per min but in the meantime we have 9 requests. 

# def fetch_data(symbol, api_key):
    # url = f"https://api.twelvedata.com/time_series?symbol={symbols}&interval=1min&apikey={api_key}&outputsize=50"
    # response = requests.get(url)
    # response.raise_for_status() # this will raise an error for bad responses
    # data = response.json()

    # if data.get('status') != 'ok':
        # raise ValueError(f'Error fetching data for {symbol}:{data.get("message", "Unknown error")}')

    # return data
```

```python
symbols = [ 'AAPL', 'MSFT', 'GOOGL', 'AMZNTSLA', 'TSLA', 'NVDA', 'NFLX', 'META', 'JPM']
```

```python

# We are using the batch endpoint to fetch all symbols in one API call.
# This avoids hitting the free-tier rate limit and keeps the workflow simple reason for the loop below and commenting out the fetch_data symbol in symbols code.
# for symbol in symbols:
    # data = fetch_data(symbol, api_key)
    # df = transform_data(data)
    # df['symbol'] = symbol
    # all_data.append(df)

# all_data

all_df = pd.DataFrame()

for symbol in symbols: # Skip symbols missing from the batch response
    if symbol not in data:
        continue

    symbol_data = data[symbol]

    # Skip symbols with no 'values' field
    if "values" not in symbol_data:
        continue

    df = transform_data(symbol_data)
    df['symbol'] = symbol
    all_data.append(df)   
```

```python
# concat - joining or stacking all the dateframes together
all_df = pd.concat(all_data, ignore_index = True)
```
